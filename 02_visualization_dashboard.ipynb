# Visualisation Dashboard
Builds a 6-panel dashboard from the summary CSVs created in `01_sales_analysis.ipynb` (run that notebook first).

| Panel | Chart | Shows |
|---|---|---|
| 1 | Pie | Sales share of each category |
| 2 | Bar | Clothing sales by sub-category |
| 3 | Bar | Electronics sales by sub-category |
| 4 | Bar | Furniture sales by sub-category |
| 5 | Pie | Top 10 states by sales |
| 6 | Pie | Top 10 customers by sales |

```python
import pandas as pd
import matplotlib.pyplot as plt

#--------------------------------------------------------
# 1. Loading Grouped Sales Data
#--------------------------------------------------------

category_sales = pd.read_csv("/Users/krutikkukde/Desktop/DATA CLEANING/category_sales.csv")
Clothing_Sales = pd.read_csv("/Users/krutikkukde/Desktop/DATA CLEANING/Clothing_Sales.csv")
Electronics_Sales = pd.read_csv("/Users/krutikkukde/Desktop/DATA CLEANING/Electronics_Sales.csv")
Furniture_Sales = pd.read_csv("/Users/krutikkukde/Desktop/DATA CLEANING/Furniture_Sales.csv")
Top10_State = pd.read_csv("/Users/krutikkukde/Desktop/DATA CLEANING/State_sales.csv")
Top10_Cust = pd.read_csv("/Users/krutikkukde/Desktop/DATA CLEANING/Top_Cust_Sales.csv")


#--------------------------------------------------------
# 2. Creating the Figure
#--------------------------------------------------------
plt.figure(figsize=(18,10))


#--------------------------------------------------------
# 3. Plot 1: Plotting Category-wise Sales (Pie Chart)
#--------------------------------------------------------
plt.subplot(2, 3, 1)

plt.pie(
    category_sales['Amount'],
    labels=category_sales['Category'],
    autopct='%1.1f%%',
    colors=['steelblue', 'lightcoral', 'mediumseagreen'],
)

plt.title("Sales Distribution by Category", pad=20)

plt.legend(
    loc="upper center",
    ncol=1,
    bbox_to_anchor=(0.5, -0.02)
)

#--------------------------------------------------------
# 3. Plot 2: Clothing Sales (Bar Chart)
#--------------------------------------------------------
plt.subplot(2, 3, 2)



plt.bar(
    Clothing_Sales['Sub-Category'],
    Clothing_Sales['Amount'],
    color=['steelblue', 'lightcoral', 'mediumseagreen', 'goldenrod', 'mediumpurple']

)

plt.title("Clothing Sales", pad=20)
plt.xlabel("Products", labelpad=10)
plt.ylabel("Sales")
plt.xticks(rotation=45, ha='right', rotation_mode='anchor')

#--------------------------------------------------------
# 4. Plot 3: Electronics Sales (Bar Chart)
#--------------------------------------------------------

plt.subplot(2, 3, 3)

plt.bar(
    Electronics_Sales['Sub-Category'],
    Electronics_Sales["Amount"],
    color=['lightcoral', 'mediumseagreen', 'goldenrod', 'mediumpurple']

)

plt.title("Electronic Sales", pad=20)
plt.xlabel("Products", labelpad=10)
plt.ylabel("Sales")

#--------------------------------------------------------
# 5. Plot 4: Furniture Sales (Bar Chart)
#--------------------------------------------------------

plt.subplot(2, 3, 4)

plt.bar(
    Furniture_Sales['Sub-Category'],
    Furniture_Sales["Amount"],
    color=['lightcoral', 'mediumseagreen', 'goldenrod', 'mediumpurple']

)

plt.title("Furniture Sales", pad=20)
plt.xlabel("Products", labelpad=10)
plt.ylabel("Sales")


#--------------------------------------------------------
# 6. Plot 5: Plotting Top 10 State wise Sales (Pie Chart)
#--------------------------------------------------------
plt.subplot(2, 3, 5)

plt.pie(
    Top10_State['Amount'],
    labels=Top10_State['State'],
    autopct='%1.0f%%',
    colors = ['steelblue', 'lightcoral', 'mediumseagreen', 'cornflowerblue', 'lightseagreen', 'salmon', 'mediumaquamarine', 'skyblue', 'plum', 'wheat']
)

plt.title("Top 10 State wise Sales", pad=20)

plt.legend(
    loc="upper center",
    ncol=3,
    bbox_to_anchor=(0.5, -0.02)
)

#--------------------------------------------------------
# 7. Plot 6: Plotting Top 10 Customes wise Sales (Pie Chart)
#--------------------------------------------------------
plt.subplot(2, 3, 6)

plt.pie(
    Top10_Cust['Amount'],
    labels=Top10_Cust['CustomerName'],
    autopct='%1.0f%%',
    colors = ['powderblue', 'peachpuff', 'palegreen', 'lavender', 'khaki', 'lightsteelblue', 'moccasin', 'palevioletred']
    
)

plt.title("Top 10 Customer", pad=20)

plt.legend(
    loc="upper center",
    ncol=3,
    bbox_to_anchor=(0.5, -0.02)
)

#--------------------------------------------------------
# 8. Layout Adjustment
#--------------------------------------------------------
plt.subplots_adjust(left=0.06, right=0.97, top=0.93, bottom=0.10, wspace=0.30, hspace=0.45)
```
